```python
import requests
```

```python
from bs4 import BeautifulSoup
```

```python
url = "https://www.scrapethissite.com/pages/forms/"
```

```python
response = requests.get(url)
```

```python
soup = BeautifulSoup(response.text, "html")
```

```python
table = soup.find("table")
```

```python
col = [dt for dt in table.find_all("th")]
col
```

```python
title = [dt.text.strip() for dt in col]
title
```

```python
import pandas as pd
```

```python
df = pd.DataFrame(columns= title)
```

```python
df
```

```python
for d in table.find_all("tr")[1:]:
    td = d.find_all("td")
    data = [data.text.strip() for data in td]

    length = len(df)
    df.loc[length] = data
```
